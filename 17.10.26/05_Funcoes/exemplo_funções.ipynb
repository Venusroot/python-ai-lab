```python
# DESAFIO 01
"""
Crie uma função que calcule a soma de diversos números.
"""
def sominha(*numeros):
    calculo=0
    for i in numeros:
        calculo+=i
    return calculo

print(sominha(2, 5, 100))
```

```python
# DESAFIO 02
"""
Faça um programa que receba um número inteiro de 1 a 10.
Crie uma função que apresente o resultado da tabuada desse número.
"""
def tabuada(num):
    for i in range(1, 11):
        print(f"{i} x {num} = {i*num}")

tabuada(int(input("Digite um nº pra ver sua tabuada: ")))
```

```python
# DESAFIO 03 (simplificado)
"""
Crie uma função que retorne a saudação de acordo com o nome da pessoa e hora do dia.
Ex.: Boa noite, Leandro!

Configuração:
06 e 12 = Bom dia
12 e 18 = Boa tarde
18 e 00 = Boa noite
00 e 05 = Boa madrugada
"""
def saudacao(nome, hora):
    match hora:
        case h if h < 0:
            print(f"Eita {nome}, essa hora é inválida!")
        case h if h < 6:
            print(f"Boa madrugada, {nome}!")
        case h if h < 12:
            print(f"Bom dia, {nome}!")
        case h if h < 18:
            print(f"Boa tarde, {nome}!")
        case h if h < 24:
            print(f"Boa noite, {nome}!")
        case _:
            print(f"Eita {nome}, essa hora é inválida!")

saudacao('Leandro', 17)
```
